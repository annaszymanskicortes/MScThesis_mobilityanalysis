```python
import csv
import collections
```

### user statistics

```python
# demographic stats
gender = [0] * 2
grade = [0] * 5
with open('user.csv', 'r') as file:
    reader = csv.DictReader(file)
    for row in reader:
        gender[int(row['gender']) - 1] += 1
        grade[int(row['grade']) - 1] += 1
print("gender distribution: ", gender)
print("grade distribution: ", grade)
```

```python
#checkin mood distribution
count = {}
count_weekdays = {}
count_weekends = {}
moods = ['happy', 'ordinary', 'sad', 'fearful', 'angry']
for mood in moods:
    count[mood] = 0
    count_weekdays[mood] = 0
    count_weekends[mood] = 0
with open('checkIns.csv', 'r') as file:
    reader = csv.DictReader(file)
    for row in reader:
        count[row['mood']] += 1
        if row['day_of_week'] in ['1', '7']:
            count_weekends[row['mood']] += 1
        else:
            count_weekdays[row['mood']] += 1
print("total mood distribution: ", count)
print("weekday mood distribution: ", count_weekdays)
print("weekend mood distribution: ", count_weekends)
```

### temporal and spatial patterns of checkins

```python
# time
count = [0] * 24
with open('checkIns.csv', 'r') as file:
    reader = csv.DictReader(file)
    for row in reader:
        count[int(row['time'][:2])] += 1
        # count[int(row['day_of_week'])- 1][int(row['time'][:2])] += 1
print("number of check-ins across 24 hours of a day:")
print(count)
```

```python
# POI categories
category = collections.Counter()
poi = collections.Counter()
with open('checkIns.csv', 'r') as file:
    reader = csv.DictReader(file)
    for row in reader:
        category[row["POI_category"]] += 1
print("number of POI categories:", len(category))
print("distribution:", dict(category))
```
